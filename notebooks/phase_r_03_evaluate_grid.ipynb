```python
import warnings; warnings.filterwarnings('ignore')
import os
# IDTxl can JVM - dat JAVA_HOME truoc khi import baseline
os.environ.setdefault('JAVA_HOME', r'C:\Program Files\Java\jdk-22')
import yaml, numpy as np, pandas as pd
from pathlib import Path

from pqrst.data.synthetic.corpus import load_corpus
from pqrst.estimators.mine.amortized import AmortizedTEEstimator
from pqrst.baselines.ksg import KSGTEEstimator
from pqrst.baselines.binning import BinningTEEstimator
from pqrst.baselines.symbolic_te import SymbolicTEEstimator
from pqrst.evaluation.grid import evaluate_estimators_on_grid, summarize_grid

BASE = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
```

## 1. Nap tap test + checkpoint T_phi

```python
test_windows = load_corpus(str(BASE / 'data' / 'interim' / 'synthetic_corpus' / 'test.npz'))
checkpoint_path = BASE / 'results' / 'checkpoints' / 'phi_amortized.pt'
est_amortized = AmortizedTEEstimator.load(str(checkpoint_path))
```

## 2. Lap dict 4 estimator

Tat ca tuan thu cung interface `BaseTEEstimator` - day la ly do vong lap danh gia khong can biet ben trong la gi.

```python
estimators = {
    'Amortized': est_amortized,
    'KSG': KSGTEEstimator(),
    'Symbolic': SymbolicTEEstimator(),
    'Binning': BinningTEEstimator()
}
```

## 3. Chay danh gia tren toan luoi

Bat `progress=True`. Estimator loi tren cua so nao thi ghi NaN + dem, **khong** de vo vong lap.

```python
raw = evaluate_estimators_on_grid(estimators, test_windows, progress=True)
out_dir = BASE / 'results' / 'tables'
out_dir.mkdir(parents=True, exist_ok=True)
raw.to_csv(out_dir / 'phase_r_grid_raw.csv', index=False)
```
