# Pha R (mo rong) — Buoc 4: Phan tich - KSG con thang o N nho tren du lieu phi tuyen khong?

So sanh truc tiep voi ket qua VAR tuyen tinh goc (`results/tables/phase_r_grid_summary.csv`).

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path

BASE = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
summary = pd.read_csv(BASE/'results'/'tables'/'phase_r2_periodic_grid_summary.csv')
summary.head()
```

## 1. Variance vs N tren du lieu periodic coupling

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=summary, x='n_samples', y='variance', hue='estimator', marker='o', errorbar=None)
plt.axvline(30, color='r', linestyle='--', label='N=30')
plt.xscale('log'); plt.yscale('log')
plt.title('Variance vs N — Periodic Coupling')
plt.legend()
plt.savefig(BASE/'results'/'figures'/'phase_r2_periodic_variance_vs_n.png', dpi=100)
plt.show()
```

## 2. Verdict — dinh luong, so sanh voi ket qua VAR tuyen tinh

```python
pivot = summary[summary['n_samples'] < 30].pivot_table(
    index=['config_name','coupling_c','noise_std','n_samples'],
    columns='estimator', values='variance').reset_index()
n_cells = len(pivot)
beats_ksg = (pivot['Amortized'] < pivot['KSG']).sum()
beats_sym = (pivot['Amortized'] < pivot['Symbolic']).sum()
beats_bin = (pivot['Amortized'] < pivot['Binning']).sum()
print(f'So o luoi N<30: {n_cells}')
print(f'Amortized thang KSG:      {beats_ksg}/{n_cells}')
print(f'Amortized thang Symbolic: {beats_sym}/{n_cells}')
print(f'Amortized thang Binning:  {beats_bin}/{n_cells}')

print('\n--- So sanh voi ket qua VAR tuyen tinh Gaussian goc: 9/30 thang KSG ---')
if n_cells > 0:
    rate = beats_ksg / n_cells
    if rate > 9/30 + 0.1:
        print('=> KSG YEU HON ro ret tren du lieu phi tuyen - xac nhan gia thuyet '
              "'KSG chi manh tren VAR Gaussian, san nha cua no'.")
    elif rate < 9/30 - 0.1:
        print('=> KSG con manh hon tren du lieu phi tuyen - bac bo gia thuyet, '
              'KSG khong dac thu cho Gaussian nhu nghi ngo.')
    else:
        print('=> Ty le tuong tu VAR tuyen tinh - loi the/bat loi cua tung phuong '
              'phap co ve khong phu thuoc dang phan phoi du lieu.')
```
