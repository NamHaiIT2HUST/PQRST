# Ước lượng Pseudo Ground Truth Transfer Entropy

Notebook này được dùng để chạy KSG trên N rất lớn (200,000) nhằm tính giá trị transfer entropy tham chiếu cho các bộ dữ liệu không có công thức đóng (VAR nonlinear và Periodic coupling).

```python
import numpy as np
import yaml
import warnings
warnings.filterwarnings("ignore")

from pqrst.data.synthetic.var_nonlinear import generate_var_nonlinear
from pqrst.data.synthetic.periodic_coupling import generate_periodic_coupling
from pqrst.baselines.ksg import KSGTEEstimator

np.random.seed(42)
```

## Khởi tạo Estimator

```python
est = KSGTEEstimator(k=4)
N_SAMPLES = 200000
```

## VAR Nonlinear

```python
x_c, y_c = generate_var_nonlinear(N_SAMPLES, a=0.5, b=0.5, c=0.6, noise_std=0.5, seed=42)
te_c = est.estimate(x_c, y_c)
print(f"VAR Nonlinear (Coupled): {te_c:.6f} nats")

x_u, y_u = generate_var_nonlinear(N_SAMPLES, a=0.5, b=0.5, c=0.0, noise_std=0.5, seed=42)
te_u = est.estimate(x_u, y_u)
print(f"VAR Nonlinear (Uncoupled): {te_u:.6f} nats")
```

## Periodic Coupling

```python
x_pc, y_pc = generate_periodic_coupling(N_SAMPLES, omega_x=0.3, omega_y=0.31, coupling_k=0.8, noise_std=0.1, seed=42)
te_pc = est.estimate(x_pc, y_pc)
print(f"Periodic Coupling (Coupled): {te_pc:.6f} nats")

x_pu, y_pu = generate_periodic_coupling(N_SAMPLES, omega_x=0.3, omega_y=0.31, coupling_k=0.0, noise_std=0.1, seed=42)
te_pu = est.estimate(x_pu, y_pu)
print(f"Periodic Coupling (Uncoupled): {te_pu:.6f} nats")
```
